```python
# 我個人會選這個, 因為這個比較好擴充
# 快捷鍵: TAB, Shift+TAB
def is_leap_year(year):
    # year = 2000
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False
```
