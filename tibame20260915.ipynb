```python
# 我個人會選這個, 因為這個比較好擴充
# 快捷鍵: TAB, Shift+TAB
def is_leap_year(year):
    # year = 2000
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False
```

```python
# d = 20241001
# # 01
# day = d %  100
# # 202410
# remain = d // 100
# # 10
# month = remain % 100
# # 2024
# year = remain // 100
# print(year, month, day)

def is_date_valid(d):
    # d = 20240229
    # 29
    day = d %  100
    # 202402
    remain = d // 100
    # 02
    month = remain % 100
    # 2024
    year = remain // 100
    # print(year, month, day)

    if year >= 1:
        if 12 >= month >= 1:
            day_per_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
            max_days = day_per_month[month-1]
            # if month == 2 and is_leap_year(year) == True
            if month == 2 and is_leap_year(year):
                max_days = max_days + 1
            if max_days >= day >= 1:
                return True
            else:
                # day invalid
                return False
        else:
            # month invalid
            return False
    else:
        # year invalid
        return False
    # return False

print(is_date_valid(20240229))
print(is_date_valid(20240230))
```

```python
def thought_2(start=4499, maxcount=10):
# start = 4499
    count = 0
    result = []
    while True:
        # 數字平方
        d= start * start
        # if is_date_valid(d) == True
        if is_date_valid(d):
            result.append(d)
            count = count + 1
        if count == maxcount:
            break
        start = start + 1
    return result

print(thought_2())
print(thought_2(maxcount=2))
print(thought_2(start=1, maxcount=5))
```
